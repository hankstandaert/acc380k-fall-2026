### If Things Get Confusing

Colab lets us run cells in almost any order. That is useful, but it can also leave us wondering what has actually run. When that happens:

1. Save your edits.
2. Use **Runtime → Restart session and run all**. The exact menu wording may vary slightly.
3. Read the first error, if one appears.
4. Check the final outputs again.

Restarting and running from the top is a good way to make sure the notebook still works. It does not, by itself, prove that every answer is right.
