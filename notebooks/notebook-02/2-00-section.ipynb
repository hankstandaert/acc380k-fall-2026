## Names in Python

A Python name can contain letters, digits, and underscores, but it cannot begin with a digit. Spaces and hyphens do not work inside a name. Python also reserves some words for its own syntax, including `and`, `or`, and `not`.

Names are case-sensitive, so `score` and `Score` are two different names. In this course, prefer descriptive lowercase words separated by underscores, such as `observation_count`.

| Valid name | Invalid name | Why the invalid name fails |
|---|---|---|
| `score_2` | `2_score` | A name cannot begin with a digit. |
| `total_score` | `total-score` | Python reads `-` as subtraction. |
| `student_count` | `student count` | A name cannot contain a space. |

Avoid using names such as `print`, `type`, `str`, `int`, or `round` for your own values. Python already uses those names for functions.


## Values and types

### What a type tells Python

A value's **type** tells Python what kind of thing the value is and which operations make sense. Quotation marks are not decoration: they create text.

For now, focus on four common types and one special marker:

- `int`: a whole number, such as `8`;
- `float`: a number with a decimal point, such as `8.0`;
- `str`: text, such as `"8"`;
- `bool`: either `True` or `False`; and
- `None`: a special value indicating that no value is currently present or available. `None` is the only value whose type is `NoneType`; it is not the same as `False`, `0`, or empty text.

## Comparisons produce Boolean values

An assignment records a value. A comparison asks a question and produces `True` or `False`.

| Operator | Question |
|---|---|
| `==` | Are the values equal? |
| `!=` | Are the values different? |
| `<`, `<=` | Is the left value less than, or less than/equal to, the right? |
| `>`, `>=` | Is the left value greater than, or greater than/equal to, the right? |

A single `=` assigns. Two equals signs, `==`, compare.


### Predict at the boundary

A measurement is exactly equal to the threshold. Predict all four Boolean results before running the cell. Pay particular attention to `>` versus `>=`.


````python
measurement = 100
threshold = 100

print("==", measurement == threshold)
print("!=", measurement != threshold)
print(">", measurement > threshold)
print(">=", measurement >= threshold)
````

## Arithmetic operators

## Keep-for-later reference

`type()`, `str()`, `int()`, and `round()` are functions. The values or expressions supplied inside their parentheses are arguments.

- Inspect a value's type: `type(value)`
- Mark that no value is currently available: `None`
- Convert to text: `str(value)`
- Convert text that represents a whole number: `int(text)`
- Assign a value: `name = expression`
- Compare values: `==`, `!=`, `<`, `<=`, `>`, `>=`
- Combine Boolean questions: `and`, `or`, `not`
- Calculate: `+`, `-`, `*`, `/`, `**`
- Round a number: `round(value, digits)`
- Add a one-line code comment: begin with `#`

Python includes many other types—lists, dictionaries, sets, tuples, and more. We will introduce structures when they solve a problem we actually need to solve.
